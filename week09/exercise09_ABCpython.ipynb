```
import pandas as pd

# Create the customers DataFrame based off the create table code I made originally
customers = pd.DataFrame(columns=['customer_id', 'name', 'state'])

#this sets the values
customers = customers.astype({
    'customer_id': 'Int64',   
    'name': 'string',         
    'state': 'string'         
})
```

```
# insert data into the dataframe
new_rows = pd.DataFrame([
    {'customer_id': 1, 'name': 'Alice', 'state': 'MO'},
    {'customer_id': 2, 'name': 'BOB',   'state': 'IL'},
    {'customer_id': 3, 'name': 'Carol', 'state': 'TX'}
])

customers = pd.concat([customers, new_rows], ignore_index=True).astype({
    'customer_id': 'Int64',
    'name': 'string',
    'state': 'string'
})
```

```
# Creates a data frame for the orders
orders = pd.DataFrame([
    {'order_id': 101, 'customer_id': 1, 'amount': 50.00},
    {'order_id': 102, 'customer_id': 2, 'amount': 75.00},
    {'order_id': 103, 'customer_id': 1, 'amount': 30.00}
])

#inner join equivilant for the customer and order ids
result_q3 = orders.merge(customers, on='customer_id', how='inner')[['name', 'order_id', 'amount']]

print(result_q3)
```

```
#Equivilant of a right join of the orders and customers
result_q4 = orders.merge(customers, on='customer_id', how='right')[['name', 'order_id']]

print(result_q4)
```

```
# defines the orders 
orders = pd.DataFrame([
    {'order_id': 101, 'customer_id': 1, 'amount': 50.00},
    {'order_id': 102, 'customer_id': 2, 'amount': 75.00},
    {'order_id': 103, 'customer_id': 1, 'amount': 30.00}
])

# Sets up the returns dataframe
returns = pd.DataFrame([
    {'return_id': 1, 'order_id': 101, 'return_date': '2024-01-15'},
    {'return_id': 2, 'order_id': 103, 'return_date': '2024-01-20'}
])

# inner join orders and returns, then essencially mimics case logic. 
result_q5 = orders.merge(returns, on='order_id', how='inner')

#This code replaces the following
# CASE WHEN r.order_id = o.order_id THEN 'True' ELSE 'False'
# After an inner join, order_id always matches — but replicating the check for accuracy
result_q5['is_return'] = result_q5['order_id'].apply(lambda x: 'True' if x else 'False')


result_q5 = result_q5[['order_id', 'return_id', 'is_return']]

print(result_q5)
```

```
# This follows an innr join orders and returns, pull back validation + return info
result_q6 = orders.merge(returns, on='order_id', how='inner')[['order_id', 'return_date', 'return_id']]

print(result_q6)
```

```
# Find customers with no orders using a LEFT JOIN + filter nulls
result_q7 = customers.merge(orders, on='customer_id', how='left')
result_q7 = result_q7[result_q7['order_id'].isna()][['name']]

print(result_q7)
```

```
# Define patients DataFrame
patients = pd.DataFrame([
    {'patient_id': 1, 'patient_name': 'John'},
    {'patient_id': 2, 'patient_name': 'Jane'},
    {'patient_id': 3, 'patient_name': 'Mike'}
])

# Define visits DataFrame
visits = pd.DataFrame([
    {'visit_id': 1001, 'patient_id': 1, 'provider_id': 1, 'visit_date': '2024-02-01'},
    {'visit_id': 1002, 'patient_id': 2, 'provider_id': 1, 'visit_date': '2024-02-05'},
    {'visit_id': 1003, 'patient_id': 1, 'provider_id': 2, 'visit_date': '2024-02-10'}
])

# Define providers DataFrame
providers = pd.DataFrame([
    {'provider_id': 1, 'provider_name': 'Dr. Smith'},
    {'provider_id': 2, 'provider_name': 'Dr. Jones'},
    {'provider_id': 3, 'provider_name': 'Dr. Lee'}
])

# Chain two inner joins: patients -> visits -> providers
result_q1 = (patients
             .merge(visits, on='patient_id', how='inner')
             .merge(providers, on='provider_id', how='inner')
             [['patient_id', 'patient_name', 'visit_date', 'visit_id', 'provider_name']]
)

print(result_q1)
```

```
# Q7 - Left join patients and visits
result_q2b = patients.merge(visits, on='patient_id', how='left')[['patient_id', 'visit_id']]

print(result_q2b)
```

```
# sets up the providers df
providers = pd.DataFrame([
    {'provider_id': 1, 'provider_name': 'Dr. Smith'},
    {'provider_id': 2, 'provider_name': 'Dr. Jones'},
    {'provider_id': 3, 'provider_name': 'Dr. Lee'}
])

# sets up the visits df
visits = pd.DataFrame([
    {'visit_id': 1001, 'patient_id': 1, 'provider_id': 1, 'visit_date': '2024-02-01'},
    {'visit_id': 1002, 'patient_id': 2, 'provider_id': 1, 'visit_date': '2024-02-05'},
    {'visit_id': 1003, 'patient_id': 1, 'provider_id': 2, 'visit_date': '2024-02-10'}
])

# Satisfies the left join for providers and visits
result_q3b = providers.merge(visits, on='provider_id', how='left')[['provider_id', 'visit_id']]

print(result_q3b)
```

```
# Checks for patients with no visits
result_q4b = patients.merge(visits, on='patient_id', how='left')
result_q4b = result_q4b[result_q4b['visit_id'].isna()][['patient_name']]

print(result_q4b)
```

```
# Redefine providers with specialty column
providers = pd.DataFrame([
    {'provider_id': 1, 'provider_name': 'Dr. Smith', 'specialty': 'Cardiology'},
    {'provider_id': 2, 'provider_name': 'Dr. Jones', 'specialty': 'Neurology'},
    {'provider_id': 3, 'provider_name': 'Dr. Lee', 'specialty': 'Cardiology'}
])

# this pulls for Patients who visited a Cardiology provider
result_q5b = (patients
              .merge(visits, on='patient_id', how='inner')
              .merge(providers, on='provider_id', how='inner')
)

# forspecialty = 'Cardiology'
result_q5b = result_q5b[result_q5b['specialty'] == 'Cardiology'][['patient_id', 'patient_name', 'visit_date', 'visit_id', 'provider_name']]

print(result_q5b)
```

```
# Define quality test in a dfe
quality_tests = pd.DataFrame([
    {'test_id': 1, 'batch_id': 1, 'result': 'pass', 'test_type': 'purity'},
    {'test_id': 2, 'batch_id': 1, 'result': 'fail', 'test_type': 'potency'},
    {'test_id': 3, 'batch_id': 2, 'result': 'pass', 'test_type': 'purity'}
])

# Define batches af
batches = pd.DataFrame([
    {'batch_id': 1, 'batch_name': 'Batch A'},
    {'batch_id': 2, 'batch_name': 'Batch B'},
    {'batch_id': 3, 'batch_name': 'Batch C'}
])

# Right join quality_tests and batches that also makes sure to filter out nulls
result_q1c = quality_tests.merge(batches, on='batch_id', how='right')

# Replaces where q.test_id IS NOT NULL
result_q1c = result_q1c[result_q1c['test_id'].notna()][['batch_id', 'result', 'test_type']]

print(result_q1c)
```

```
# right join quality_tests and batches
result_q2c = quality_tests.merge(batches, on='batch_id', how='right')[['batch_id', 'result', 'test_type']]

print(result_q2c)
```

```
#another df set up for deviations
deviations = pd.DataFrame([
    {'deviation_id': 1, 'batch_id': 1, 'deviation_type': 'temperature'},
    {'deviation_id': 2, 'batch_id': 2, 'deviation_type': 'humidity'},
    {'deviation_id': 3, 'batch_id': 4, 'deviation_type': 'contamination'}
])

# Left join that filters on fail. 
result_3c = (quality_tests
             .merge(batches, on='batch_id', how='left', suffixes=('_q', '_b'))
             .merge(deviations, on='batch_id', how='right', suffixes=('', '_d'))
)
result_3c = result_3c[result_3c['result'] == 'fail'][['batch_id', 'result', 'deviation_id']]

print(result_3c)
```

```
# Inner join for quality tests, batches, and deviations
result_4c = (quality_tests
             .merge(batches, on='batch_id', how='inner')
             .merge(deviations, on='batch_id', how='inner')
             .groupby('batch_id')
             .agg(deviation_count=('deviation_id', 'count'), test_count=('test_id', 'count'))
             .reset_index()
)
#batch idea is used for counting
print(result_4c)
```

```
#this is the anti join using a left join as the base. 
result_5c = batches.merge(deviations, on='batch_id', how='left')
result_5c = result_5c[result_5c['deviation_id'].isna()][['batch_id']]

print(result_5c)
```
