```
%%sql -r dataframe_20
--Q3
SELECT b.batch_id, q.batch_id, d.batch_id, q.result, d.deviation_id
FROM quality_tests q
LEFT JOIN batches b ON b.batch_id = q.batch_id -- left join for one side of the q table
RIGHT JOIN deviations d ON d.batch_id = q.batch_id --right join for the other side
WHERE q.result = 'fail' -- added this in as a filter. Has to come in at the end
```

```
%%sql -r dataframe_21
--Q4
--this question is asking me to count based on the batch_id and I seemingly need to d oa deviation count for B1. Knowing this I will need to isolate it and then add up its values from across tables.
SELECT b.batch_id, COUNT(d.deviation_id) AS deviation_count, COUNT(q.test_id) AS test_count --this aggregates the data for the counts requested
FROM quality_tests q
INNER JOIN batches b ON b.batch_id = q.batch_id
INNER JOIN deviations d ON q.batch_id = d.batch_id
GROUP BY b.batch_id
```
